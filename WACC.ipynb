**WACC (Weighted Average Cost of Capital)**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#WACC = (Weight of Equity × Cost of Equity) + (Weight of Debt × Cost of Debt × (1 − Tax Rate))
```

```python
rf = 4.5/100
market_premium = 5.5/100
beta = 0.92
tax_rate = 17.9/100
market_cap = 450
total_debt = 16.26
interest_expense = 0.575
```

```python
growth_rate = 0.1
terminal_rate = 2.5/100
fcf = 14.33
cash = 8.59
```

```python
cost_of_equity = rf + (beta*market_premium)
cost_of_debt = interest_expense/total_debt
```

```python
weight_of_equity = market_cap / (market_cap + total_debt)
weight_of_debt = total_debt / (market_cap + total_debt)
```

```python
wacc = (weight_of_equity*cost_of_equity) + (weight_of_debt*(cost_of_debt*(1-tax_rate)))
wacc
```

```python
fcf_list = []
projected_fcf = fcf
for year in range(1,6):
  projected_fcf = fcf*(1 + growth_rate)**(year)
  fcf_list.append(projected_fcf)
  print(f"Projected fcf is {projected_fcf} billion dollars")
```

```python
pv_list = []
pv = fcf

for year in range(1,6):
  pv = fcf_list[year-1]/(1 + wacc)**(year)
  pv_list.append(pv)
  print(f"Projected pv is {pv} billion dollars")
```

```python
terminal_value = fcf_list[-1]*(1 + terminal_rate)/(wacc - terminal_rate)
terminal_value
```

```python
pv_of_tv = terminal_value/(1 + wacc)**5
pv_of_tv
```

```python
enterprise_value = sum(pv_list) + pv_of_tv
enterprise_value
```

```python
equity_value = enterprise_value - total_debt + cash
equity_value
```
